### Why use it?

- Eliminates boilerplate key checks
- Ideal for counting, grouping, and accumulation

## TruncatedSVD for TF-IDF (LSA)

`TruncatedSVD` performs dimensionality reduction on sparse TF-IDF matrices and implements **Latent Semantic Analysis (LSA)**.

### Why use it?
- TF-IDF matrices are large and sparse
- PCA does not work well on sparse text data
- TruncatedSVD scales efficiently and preserves semantic structure

### What it does
Given a TF-IDF matrix \(X\), it approximates:
\[
X \approx U_k \Sigma_k V_k^T
\]
where each document is represented in a low-dimensional **topic space**.

### Example
```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

vectorizer = TfidfVectorizer(max_features=20000)
X_tfidf = vectorizer.fit_transform(docs)

svd = TruncatedSVD(n_components=100, random_state=42)
X_topics = svd.fit_transform(X_tfidf)

# To get the V_k matrix (word vectors)
svd.components_ # Each row is a topic, shape (k, vocab_size)

## Accessing Word Vectors from TruncatedSVD (LSA)

After applying `TruncatedSVD` to a TF-IDF matrix, the term-topic matrix
\(V_k^\top\) is available via `svd.components_`.

### Key facts
- `svd.components_` has shape `(n_components, vocab_size)`
- Each column corresponds to a word
- Transposing gives word vectors in topic space

### Example
```python
terms = vectorizer.get_feature_names_out()

Vt = svd.components_      # V_k^T
word_vectors = Vt.T       # shape: (vocab_size, n_components)

# vector for a word
word = "data"
idx = list(terms).index(word)
vec = word_vectors[idx]
