```python
import pandas as pd
import urllib.parse
```

```python
PATH = "../Data/"
PATH_DATA = PATH + "wikispeedia_paths-and-graph/"
PATH_ARTICLES_PLAINTEXT = PATH + "wikispeedia_articles_plaintext/"
PATH_ARTICLES_HTML = PATH + "wikispeedia_articles_html/"
```

```python
def load_tsv_files(file_path, sep="\t"):
    column_names = None
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            if line.startswith("# FORMAT:"):
                column_names = line.replace("# FORMAT:", "").strip().split()
                break
    if column_names is None:
        raise ValueError(f"No '# FORMAT:' line found in {file_path}")

    df = pd.read_csv(
        file_path,
        sep=sep,
        comment='#', 
        header=None,
        names=column_names
    )
    return df
```

```python
files = ["articles.tsv", "categories.tsv", "links.tsv", "paths_finished.tsv", "paths_unfinished.tsv"]
data = {fname.split(".")[0]: load_tsv_files(PATH_DATA + fname) for fname in files}
```

```python
articles = data["articles"]
articles['article'] = articles['article'].apply(urllib.parse.unquote)
```

```python
categories = data["categories"]
categories['article'] = categories['article'].apply(urllib.parse.unquote)
categories['category'] = categories['category'].str.replace('subject.', '', regex=False).str.split('.')
```

```python
links = data["links"]
links['linkSource'] = links['linkSource'].apply(urllib.parse.unquote)
links['linkTarget'] = links['linkTarget'].apply(urllib.parse.unquote)
```

```python
paths_finished = data["paths_finished"]
paths_finished['path'] = paths_finished['path'].apply(lambda x: [urllib.parse.unquote(article) for article in x.split(',')])
```

```python
paths_unfinished = data["paths_unfinished"]
paths_unfinished['path'] = paths_unfinished['path'].apply(lambda x: [urllib.parse.unquote(article) for article in x.split(',')])
paths_unfinished['target'] = paths_unfinished['target'].apply(urllib.parse.unquote)
```
